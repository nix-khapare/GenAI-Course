```python
pragraph ="""Machine learning is powerful.
It learns from data.
Models make predictions.
Algorithms find patterns.
Data quality matters a lot.
Training takes time and resources.
Accuracy improves with better features.
Overfitting is a common problem.
Regularization helps to fix it.
Testing checks the model’s performance.
Machine learning keeps evolving fast.
"""
```

```python
# First, download required NLTK data
import nltk
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('stopwords')

# Import required modules
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer

# Create instances of stemmers and lemmatizer
porter_stemmer = PorterStemmer()
snowball_stemmer = SnowballStemmer('english')
lemmatizer = WordNetLemmatizer()

# Tokenize paragraph into sentences once
sentences = nltk.sent_tokenize(pragraph)
stopword = set(stopwords.words('english'))

# 1. Porter Stemmer
print("Porter Stemmer Results:")
porter_results = []
for sentence in sentences:
    words = word_tokenize(sentence)
    words = [porter_stemmer.stem(word) for word in words if word.lower() not in stopword]
    porter_results.append(' '.join(words))
    print(' '.join(words))

print("\nSnowball Stemmer Results:")
# 2. Snowball Stemmer
snowball_results = []
for sentence in sentences:
    words = word_tokenize(sentence)
    words = [snowball_stemmer.stem(word) for word in words if word.lower() not in stopword]
    snowball_results.append(' '.join(words))
    print(' '.join(words))

print("\nLemmatizer Results:")
# 3. WordNet Lemmatizer
lemmatizer_results = []
for sentence in sentences:
    words = word_tokenize(sentence)
    words = [lemmatizer.lemmatize(word) for word in words if word.lower() not in stopword]
    lemmatizer_results.append(' '.join(words))
    print(' '.join(words))
```
