LIMITIZATION FOR TEXT PROCESSING
Lemmatization is the process of reducing a word to its base or root form, known as a lemma. Unlike stemming, which simply removes prefixes and suffixes from words, lemmatization takes into account the context and meaning of the word to determine its correct base form. It uses a vocabulary and morphological analysis of words to achieve this.

It can be used in problem statement like    
1. Classification problem (ex product has positive or negative review)
2. Clustering problem (ex group similar news article together)
Some common lemmatization techniques include:
1. Rule-based lemmatization: This technique uses a set of predefined rules to identify the base form of a word based on its suffixes and prefixes.
2. Dictionary-based lemmatization: This technique uses a dictionary or lexicon to look up the base form of a word.
3. Machine learning-based lemmatization: This technique uses machine learning algorithms to learn the patterns and relationships between words and their base forms from a large corpus of text.
4. Hybrid lemmatization: This technique combines multiple approaches, such as rule-based and dictionary-based methods, to improve the accuracy of lemmatization.
5. Neural network-based lemmatization: This technique uses deep learning models, such as recurrent neural networks (RNNs) or transformers, to capture the context and semantics of words for more accurate lemmatization.
# Example of Lemmatization using NLTK library in Python
```python
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()
words = ["running", "jumps", "easily", "fairly"]
lemmatized_words = [lemmatizer.lemmatize(word) for word in words]
print(lemmatized_words)
```
Output:
['running', 'jump', 'easily', 'fairly']
