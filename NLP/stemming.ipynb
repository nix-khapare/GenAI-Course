RegexpStemmer class

NLTK has regexpstemmer class with the help of which we can easily implement regular expression stemmer algoritham. It basically takes a single regular expression and removes any prefix or sufix that matches the expression. 

```python
from nltk.stem import RegexpStemmer
reg_stemmer = RegexpStemmer('ing$|s$|s$|able$', min=4)
## regular expression saying wherever ing, s, e, able at the end remove that word
reg_stemmer.stem("eating")  


```

```python

reg_stemmer.stem("intigrating")  ##intigrat
reg_stemmer.stem("family") ## family
```

```python
reg_stemmer = RegexpStemmer('ing|s$|s$|able$', min=4)
## regular expression saying wherever ing, s, e, able at the end remove that word
reg_stemmer.stem("ingeating")  

```

1.Snowball stemmer : It gives better performance compare to porter stemmer
2. SnowballStemmer  provide different different languages

```python
from nltk.stem import SnowballStemmer
snowball_stemmer = SnowballStemmer("english")
snowball_stemmer.stem("congratulations")

for word in words:
    print(word,"------>",snowball_stemmer.stem(word))
```
