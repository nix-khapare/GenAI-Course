Stopwords:
Stopwords are commonly used words in a language that do not carry significant meaning and are often filtered out in natural language processing (NLP) tasks. Examples of stopwords in English include "is," "the," "and," "in," and "to." These words are frequently used in sentences but do not contribute much to the overall understanding of the text.

```python
from nltk.stem import PorterStemmer
```

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
```

```python
#this will return alll the stopwords present for english language
#You can create your own list of stopwords as per your requirement
#Different languages have different stopwords
stopwords.words('english')
```

```python
paragraph = """Machine learning is powerful.
It learns from data.
Models make predictions.
Algorithms find patterns.
Data quality matters a lot.
Training takes time and resources.
Accuracy improves with better features.
Overfitting is a common problem.
Regularization helps to fix it.
Testing checks the model’s performance.
Machine learning keeps evolving fast."""
```

```python
##
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
sentences =nltk.sent_tokenize(paragraph)
```

```python
type(sentences)
```

```python
##first we will apply stopwords and filter  then apply steming
nltk.sent_tokenize(paragraph)

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])

    ##Now i have to stem thos words which not present in stopwords
    stopword = set(stopwords.words('english'))
    words = [stemmer.stem(word) for word in words if word not in stopword]
    
    ## now convert that words into sentences
    sentences[i] = ' '.join(words)
    print(sentences[i])
    


```

```python
##as you no porterstemmer dose not give proper root word we can use lemmatization

nltk.sent_tokenize(paragraph)

from nltk.stem import SnowballStemmer
stemmer= SnowballStemmer('english')

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    stopword = set(stopwords.words('english'))

    word = [ stemmer.stem(word) for word in words if word not in stopword]
    print(' '.join(words))

```

```python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    ##Now i have to stem thos words which not present in stopwords
    stopword = set(stopwords.words('english'))
    words = [stemmer.stem(word) for word in words if word.lower() not in stopword]
    
    ## now convert that words into sentences
    print(' '.join(words))
    
```

```python
##Do with limitization because words are stem correctly

from nltk.tokenize import word_tokenize
nltk.sent_tokenize(paragraph)

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    ##Now i have to stem thos words which not present in stopwords
    stopword = set(stopwords.words('english'))
    words = [lemmatizer.lemmatize(word) for word in words if word.lower() not in stopword]
    print(' '.join(words)) 
```
